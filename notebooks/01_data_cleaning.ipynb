```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx", dtype={"InvoiceNo": str, "StockCode": str})
print(df.columns)
df.head()
```

```python
print(df.shape)
print(df.columns.tolist())
df.dtypes
```

```python
print(df.isna().sum())
df.describe()
```

```python
print("Rows:", len(df))
print("Missing CustomerID:", df["CustomerID"].isna().sum())
print("Cancelled invoices (start with C):", df["InvoiceNo"].str.startswith("C").sum())
print("Quantity <= 0:", (df["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (df["UnitPrice"] <= 0).sum())
print("Exact duplicate rows:", df.duplicated().sum())
print("Date range:", df["InvoiceDate"].min(), "->", df["InvoiceDate"].max())
```

```python
print("Before cleaning:", len(df))

df = df[df["Quantity"] > 0]
df = df[df["UnitPrice"] > 0]
df = df[df["CustomerID"].notna()]
df = df[~df["InvoiceNo"].str.startswith("C")]

#CustomerID as text without ".0" (needed for MD5 later)
df["CustomerID"] = df["CustomerID"].astype(int).astype(str)

print("After cleaning:", len(df))
```

```python
# Rename columns to the project names (Section 5.1)
df = df.rename(columns={
    "InvoiceNo": "transaction_id",
    "StockCode": "product_id",
    "Description": "product_name",
    "Quantity": "quantity",
    "InvoiceDate": "event_time",
    "UnitPrice": "unit_price",
    "CustomerID": "customer_id",
    "Country": "country",
})
df.head()
```
